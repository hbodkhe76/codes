### Telecom Customer Churn Data Preprocessing

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
df = pd.read_csv("telecom_customer_churn.csv")

print("First 5 rows:")
display(df.head())

print("\nDataset Shape:")
print(df.shape)
```

```python
print("\nDataset Information:")
df.info()
```

```python
print("\nData Types:")
print(df.dtypes)
```

### 3. Explore the dataset

```python
print("\nSummary Statistics:")
df.describe()
```

```python
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
print("\nDuplicate Values")
print(df.duplicated().sum())
```

```python
# Fill missing values
numeric_cols = df.select_dtypes(include=np.number).columns
for col in numeric_cols:
    df[col] = df[col].fillna(df[col].median())

categorical_cols = df.select_dtypes(include='object').columns
for col in categorical_cols:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(df[col].mode()[0])

print("\nMissing values after handling:")
print(df.isnull().sum())
```

```python
# Remove spaces and standardize text values
for col in df.select_dtypes(include='object').columns:
    df[col] = df[col].astype(str).str.strip().str.lower()

print("\nCategorical values standardized.")
```

```python
# Convert identifiers to string and numeric columns to numeric
df["customerID"] = df["customerID"].astype(str)

numeric_columns = ["SeniorCitizen", "tenure", "MonthlyCharges", "TotalCharges"]
for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")

print("\nData types after conversion:")
print(df.dtypes)
```

```python
numeric_cols = df.select_dtypes(include=np.number).columns

for col in numeric_cols:
    if col == "SeniorCitizen":
        continue
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR
    df[col] = df[col].clip(lower_limit, upper_limit)

print("\nOutliers handled using IQR method.")
```

```python
# Prepare the dataset for modeling using the actual CSV columns
# No extra feature engineering is added beyond the available dataset fields.

print("\nDataset ready for model preparation.")
```

```python
# Remove customer ID because it is only an identifier
# Remove target column before encoding

df_model = df.drop(columns=["customerID", "Churn"])

df_encoded = pd.get_dummies(df_model, drop_first=True, dtype=int)

print("\nEncoded dataset shape:")
print(df_encoded.shape)
```

```python
# Churn contains the target information
print("\nChurn values:")
print(df["Churn"].value_counts())

# Create binary churn target
df_encoded["Churn"] = df["Churn"].str.lower().map({"yes": 1, "no": 0})

# Separate X and y
X = df_encoded.drop(columns=["Churn"])
y = df_encoded["Churn"]
```

```python
scaler = StandardScaler()
X_scaled = pd.DataFrame(scaler.fit_transform(X), columns=X.columns)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.20, random_state=42
)

print("\nTraining data:", X_train.shape)
print("Testing data:", X_test.shape)
```

```python
df.to_csv("Telecom_Customer_Churn_Cleaned.csv", index=False)
print("\nCleaned dataset exported successfully!")

df_encoded.to_csv("Telecom_Customer_Churn_Encoded.csv", index=False)
print("Encoded dataset exported successfully!")
```
